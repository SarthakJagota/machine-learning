```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('anomaly.csv')
df.head()
```

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.show()
```

```python
from sklearn.ensemble import IsolationForest
a=IsolationForest(contamination=0.1)
a.fit(df)
predictions=a.predict(df)
predictions
```

```python
index=np.where(predictions < 0)
index
```

```python
x=df.values
```

```python
index=np.where(predictions < 0)
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[index,0],x[index,1],edgecolors='r')
plt.show()
```
